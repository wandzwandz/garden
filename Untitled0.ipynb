```python
import requests

# Public ECan MapServer layer tracking rainfall sites
url = "https://gis.ecan.govt.nz/arcgis/rest/services/Public/WaterQualityandMonitoring/MapServer/5/query"

params = {
    'where': "SITENAME LIKE '%Clough Memorial%'",
    'outFields': 'SITENAME,LAST_SAMPLE,LAST_HOUR,RAIN_TODAY,RAIN_1_DAY_AGO,RAIN_5_DAYS_AGO',
    'f': 'json'
}

try:
    response = requests.get(url, params=params)
    response.raise_for_status()
    features = response.json().get('features', [])

    if features:
        attrs = features[0]['attributes']
        print(f"--- Live Data: {attrs['SITENAME']} ---")
        print(f"Last Sample Recieved: {attrs['LAST_SAMPLE']}")
        print(f"Rainfall Last Hour:   {attrs['LAST_HOUR']} mm")
        print(f"Rainfall Today:       {attrs['RAIN_TODAY']} mm")
        print(f"Rainfall Yesterday:   {attrs['RAIN_1_DAY_AGO']} mm")
    else:
        print("Site matching 'Clough Memorial' not found.")
except Exception as e:
    print(f"Failed to fetch public GIS data: {e}")
```
